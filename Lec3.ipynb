STRING

```python
# empty string
st=""
print(len(st))

st="anand"
# capitalize the letters in string
l=st.upper()
print(l)

#change case lower(),upper(),title(),capitalize()
x="meWoWm KC"
print(x.lower())
print(x.upper())
# title only make the first letter upper case it lowers the other character after space characters are affected in the same way
print(x.title())
# capitalize only make the first letter upper case it lowers the other character and after space characters are lowerd
print(x.capitalize())

```

```python
st="  Rewas  "
ls=" khatri "
# to remove sapce from first and last part of string (strip)
# st.strip(character which sould be removed by default it is space)
print(st+ls)
first=st.strip()
second=ls.strip()
print(first+second)
# there is also left strip(x.lstrip()) and right strip (x.rstrip())
```

```python
a=" today "
b=" weather "
c="is GooD "
# write it in proper format
d=a.strip()
x=d+b+c
l=x.capitalize()
print(l)
# replace good with bad
print(l.replace("good","bad"))# l.replace(old string, new string)
# Join list with space separator
words = ["Hello", "world"]
result = " ".join(words)
print(result)

# find index
print(l.find("good"))
```
